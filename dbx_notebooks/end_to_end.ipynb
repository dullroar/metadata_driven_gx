# The metadata-driven GX pipeline

This notebook runs the complete `demo_weather` example: expectation metadata in CSV files becomes Great Expectations suites, those suites validate tabular data, and the results are flattened into readable CSV files.

The notebook calls the same `end_to_end.run_pipeline()` function used by the command-line interface. There is no second implementation hidden here.

```python
from pathlib import Path
import sys

import pandas as pd

# Find the checkout even when Jupyter starts with another working directory.
candidates = [Path.cwd(), *Path.cwd().parents]
ROOT = next(path for path in candidates if (path / 'end_to_end.py').exists())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import end_to_end

CONFIG_FILE = ROOT / 'configs' / 'demo_weather.yaml'
OUTPUT_DIR = ROOT / 'environments' / 'demo_weather' / 'output'
DATA_DOCS = ROOT / 'environments' / 'demo_weather' / 'gx' / 'uncommitted' / 'data_docs' / 'local_site' / 'index.html'
print(f'Repository: {ROOT}')
print(f'Config: {CONFIG_FILE}')
```

## Run the pipeline

`CLEAN = False` keeps the example's existing generated artifacts. Change it to `True` only when you deliberately want a from-scratch run.

```python
CLEAN = False

end_to_end.run_pipeline(config_file=CONFIG_FILE, clean=CLEAN)
print('Pipeline complete.')
```

## Read the results

Summary files contain one row per expectation. Detail files expand failing rows so you can inspect the offending values and natural keys.

```python
summary = pd.read_csv(OUTPUT_DIR / 'CUSTOMERS' / 'CUSTOMERS_summary.csv')
summary[['table', 'expectation_type', 'column', 'success', 'unexpected_count']].head(10)
```
